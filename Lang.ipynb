## Loading the data.

```python
import os 
import dotenv 
from langchain_community.document_loaders import TextLoader
from dotenv import load_dotenv
```

```python
load_dotenv()
print(OPEN_AI_API_KEY)
#print(document) 
```

```python
from langchain_community.document_loaders import PyPDFLoader
pdf = PyPDFLoader("ai_agents.pdf") 
document = pdf.load() 
```

## Converting data to chunks

```python
# Transforming the loaded document into chunks
from langchain.text_splitter import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=20)
doc = text_splitter.split_documents(document)  
print(doc[:5])   
```

## Transforming text to vector for storing

```python
from langchain_community.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma

embedding = OpenAIEmbeddings(api_key=OPEN_AI_API_KEY)
db = Chroma.from_documents(doc, embedding)

#db is our vector store containing the document chunks. 
```

## Retrieving the data from vector store for answering the query

```python
# Retrieving the most similar documents to a query from the vector store
query = "What is an AI agent?"`
results= db.similarity_search(query)
raw_text= results[0].page_content
cleaned_text= "".join(raw_text.split("\n"))
cleaned_text
```

### Adding LLM to for the query answering.

```python
from langchain_openai import ChatOpenAI

# Use the GPT-4.0 turbo model
OPEN_AI_API_KEY = os.getenv("OPEN_AI_API_KEY")
LLM = ChatOpenAI(model="gpt-4-turbo",api_key=OPEN_AI_API_KEY)

```

```python
#Chat prompt template
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""Answer the following question based only on the provided context.
                                                   Think step by step before providing a detailed answer. 
                                                   I will tip you $1000 if the user finds it helpful.
                                                   <context>
                                                   {context}
                                                   </context>
                                                   Question: {question}""")

```

## Creating the chain and stuff document. 

```python
# Creating the stuff document i.e. Combining multiple documents into one
from langchain.chains.combine_documents import create_stuff_documents_chain
chain= create_stuff_documents_chain(LLM,prompt)
```

## Adding a Retriever to the chain

```python
# What is it -A retriever is an interface that returns documents given an unstructured query.
# Retrievers accept a string query as input and return a list of Document's as output. 
#Vector stores can be used as the backbone of a retriever, but there are other types of retrievers as well.
retriever = db.as_retriever()
```
